## 📥 Step 1: Load the Food-101 Dataset

We'll utilize TensorFlow Datasets (TFDS) to load the Food-101 dataset, which comprises 101 food categories with a total of 101,000 images.

```python

````python
!pip install tensorflow tensorflow-datasets
````

````python
import tensorflow_datasets as tfds
import tensorflow as tf
````

````python
# Load the dataset
(ds_train, ds_test), ds_info = tfds.load(
    'food101',
    split=['train', 'validation'],
    shuffle_files=True,
    as_supervised=True,
    with_info=True,
)
````

````python
# Display dataset information
print(ds_info)
````

```

*Note:* The Food-101 dataset is available through TensorFlow Datasets. For more details, you can visit the [TensorFlow Datasets catalog](https://www.tensorflow.org/datasets/catalog/food101).

## 🧹 Step 2: Preprocess the Data

We'll normalize the images and resize them to a consistent size suitable for our models.

```python

````python
# Define image size and batch size
IMG_SIZE = 128
BATCH_SIZE = 32
````

````python
# Function to normalize and resize images
def preprocess(image, label):
    image = tf.image.resize(image, [IMG_SIZE, IMG_SIZE])
    image = tf.cast(image, tf.float32) / 255.0  # Normalize to [0,1]
    return image, label
````

````python
# Apply preprocessing to the datasets
ds_train = ds_train.map(preprocess, num_parallel_calls=tf.data.AUTOTUNE)
ds_train = ds_train.cache().shuffle(1000).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

ds_test = ds_test.map(preprocess, num_parallel_calls=tf.data.AUTOTUNE)
ds_test = ds_test.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
````

````python
import matplotlib.pyplot as plt
````

````python
class_names = ds_info.features['label'].names

for images, labels in ds_train.take(1):
    plt.figure(figsize=(10, 10))
    for i in range(9):
        plt.subplot(3, 3, i + 1)
        plt.imshow(images[i])
        plt.title(class_names[labels[i]])
        plt.axis("off")
````

## 🧠 Step 3: Build and Train the ANN Model

We'll construct a simple ANN model for image classification.

```python

````python
from tensorflow.keras import layers, models
````

Here is the diagram illustrating the structure of your ANN model. It shows the flow of data through the layers:

image.png

- The Flatten layer takes the input image of size (128, 128, 3) and flattens it into a 1D vector.

- The first Dense layer has 512 neurons, uses the ReLU activation funct
ion, and learns complex features from the input.

- Dropout is applied to randomly "drop" some of the neurons during training to prevent overfitting.

- The final Dense layer has 101 neurons (for 101 classes) and uses the softmax activation to output probabilities for each class.

````python
# Define the ANN model
ann_model = models.Sequential([
    layers.Flatten(input_shape=(IMG_SIZE, IMG_SIZE, 3)),
    layers.Dense(512, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(101, activation='softmax')
])
````

````python
# Compile the model
ann_model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
````

````python
# Train the model
ann_history = ann_model.fit(ds_train, epochs=10, validation_data=ds_test)
````

## 🧠 Step 4: Build and Train the CNN Model

Next, we'll build a CNN model, which is typically more effective for image classification tasks.

```python

🧠 CNN Model Breakdown:
🔹 Input Layer
Takes images of shape (128, 128, 3) — that’s height, width, and 3 color channels (RGB).

🧱 Convolution + Pooling Layers (These extract features like edges, textures, patterns)
🌀 Conv2D(32, (3,3), activation='relu')
Applies 32 filters of size 3x3 to detect features.

ReLU introduces non-linearity (so the model can learn complex patterns).

🧼 MaxPooling2D(2,2)
Downsamples the image by taking the maximum value in 2x2 patches.

Reduces size → faster training, less overfitting.

🌀 Conv2D(64, (3,3), activation='relu')
Another set of filters — now 64 — learning deeper features.

🧼 MaxPooling2D(2,2)
Further reduces size and retains important information.

🌀 Conv2D(128, (3,3), activation='relu')
Learns even more complex features from the previous layer.

🧼 MaxPooling2D(2,2)
Final downsampling before flattening.

🧹 Flatten
Converts the 3D output into a 1D vector (so it can go into dense layers).

🔋 Dense(512, activation='relu')
A fully connected layer with 512 neurons.

Acts like a decision-maker based on the extracted features.

🚫 Dropout(0.5)
Randomly drops 50% of neurons during training.

Prevents overfitting (so it doesn’t memorize training data).

📊 Dense(101, activation='softmax')
Output layer with 101 classes (one for each food type).

Softmax gives a probability for each class — the highest one is the model's guess.

🥡 In Simple Terms:
Imagine your model as a food detective 🍽️:

It looks at the image in pieces (Conv layers).

It focuses only on the important parts (Pooling).

It remembers patterns (Dense).

It makes a guess about the food (Softmax output).

image.png

````python
# Define the CNN model
cnn_model = models.Sequential([
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(IMG_SIZE, IMG_SIZE, 3)),
    layers.MaxPooling2D(2,2),
    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2),
    layers.Conv2D(128, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2),
    layers.Flatten(),
    layers.Dense(512, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(101, activation='softmax')
])
````

````python
# Compile the model
cnn_model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
````

````python
# Train the model
cnn_history = cnn_model.fit(ds_train, epochs=10, validation_data=ds_test)
````

```

## 📊 Step 5: Evaluate and Compare the Models

We'll evaluate both models on the test dataset and compare their performance.

```python

````python
# Evaluate the ANN model
ann_loss, ann_accuracy = ann_model.evaluate(ds_test)
print(f'ANN Test Accuracy: {ann_accuracy:.4f}')
````

````python
# Evaluate the CNN model
cnn_loss, cnn_accuracy = cnn_model.evaluate(ds_test)
print(f'CNN Test Accuracy: {cnn_accuracy:.4f}')
````
